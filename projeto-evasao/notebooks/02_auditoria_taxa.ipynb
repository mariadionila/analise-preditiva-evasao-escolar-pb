```python

# CÉLULA 1 — Inspecionar as taxas de rendimento de 2025

from pathlib import Path
import pandas as pd

# Localizar a pasta do projeto
RAIZ = Path.cwd()

if RAIZ.name == "notebooks":
    RAIZ = RAIZ.parent

arquivo_taxas = (
    RAIZ / "dados-brutos" / "tx_rend_escolas_2025.xlsx"
)

if not arquivo_taxas.exists():
    raise FileNotFoundError(
        f"Arquivo não encontrado: {arquivo_taxas}\n"
        "Confira a pasta dados-brutos."
    )

# Listar as abas do Excel
xls = pd.ExcelFile(arquivo_taxas)

print("Abas disponíveis:")
print(xls.sheet_names)

# Inspecionar as primeiras linhas de cada aba
for aba in xls.sheet_names:
    dados_aba = pd.read_excel(
        arquivo_taxas,
        sheet_name=aba,
        nrows=5
    )

    print(f"\n=== Aba: {aba} ===")
    print("Colunas:")
    print(dados_aba.columns.tolist())

    display(dados_aba.head())
```

```python

# CÉLULA 2 — Ler o Excel com o cabeçalho correto

arquivo_taxas = RAIZ / "dados-brutos" / "tx_rend_escolas_2025.xlsx"

df_taxas = pd.read_excel(
    arquivo_taxas,
    sheet_name="ESCOLAS",
    header=4
)

# Remover colunas totalmente vazias
df_taxas = df_taxas.dropna(axis=1, how="all")

# Remover espaços dos nomes das colunas
df_taxas.columns = (
    df_taxas.columns
    .astype(str)
    .str.strip()
)

print("Dimensões:", df_taxas.shape)

print("\nColunas disponíveis:")
for coluna in df_taxas.columns:
    print("-", coluna)

print("\nPrimeiras linhas:")
display(df_taxas.head())

print("\nValores ausentes por coluna:")
display(
    df_taxas.isna().sum()
    .sort_values(ascending=False)
    .to_frame("valores_ausentes")
)
```

```python

# CÉLULA 3 — Corrigir cabeçalho e limpar linhas iniciais

arquivo_taxas = RAIZ / "dados-brutos" / "tx_rend_escolas_2025.xlsx"

# A linha com os nomes técnicos é a linha de índice 3
df_taxas = pd.read_excel(
    arquivo_taxas,
    sheet_name="ESCOLAS",
    header=3
)

# Remover colunas sem nome técnico
df_taxas = df_taxas.loc[
    :,
    ~df_taxas.columns.astype(str).str.startswith("Unnamed")
].copy()

# Remover espaços dos nomes das colunas
df_taxas.columns = (
    df_taxas.columns.astype(str).str.strip()
)

# Remover linhas sem código da entidade escolar
if "CO_ENTIDADE" in df_taxas.columns:
    df_taxas = df_taxas[
        df_taxas["CO_ENTIDADE"].notna()
    ].copy()

print("Dimensões após ajuste:", df_taxas.shape)

print("\nColunas disponíveis:")
for coluna in df_taxas.columns:
    print("-", coluna)

print("\nColunas relacionadas a abandono:")
print([
    coluna for coluna in df_taxas.columns
    if "ABAND" in coluna.upper()
])

print("\nAmostra:")
display(df_taxas.head())
```

```python

# CÉLULA 4 — Ler corretamente as taxas de rendimento

arquivo_taxas = RAIZ / "dados-brutos" / "tx_rend_escolas_2025.xlsx"

df_taxas = pd.read_excel(
    arquivo_taxas,
    sheet_name="ESCOLAS",
    header=8
)

# Remover colunas totalmente vazias
df_taxas = df_taxas.dropna(axis=1, how="all")

# Remover espaços dos nomes
df_taxas.columns = df_taxas.columns.astype(str).str.strip()

# Manter somente linhas que identificam uma escola
df_taxas = df_taxas[
    df_taxas["CO_ENTIDADE"].notna()
].copy()

print("Dimensões:", df_taxas.shape)

print("\nColunas de identificação:")
display(df_taxas[
    [
        "NU_ANO_CENSO",
        "SG_UF",
        "CO_MUNICIPIO",
        "NO_MUNICIPIO",
        "CO_ENTIDADE",
        "NO_ENTIDADE"
    ]
].head())

print("\nColunas de taxas do ensino médio:")
print([c for c in df_taxas.columns if "MED" in c])

print("\nAmostra das taxas do ensino médio:")
display(
    df_taxas[
        ["NU_ANO_CENSO", "SG_UF", "CO_ENTIDADE",
         "1_CAT_MED", "2_CAT_MED", "3_CAT_MED"]
    ].head()
)
```

```python

# CÉLULA 5 — Filtrar escolas da Paraíba

# Garantir que os códigos de identificação tenham tipos consistentes
df_taxas["SG_UF"] = df_taxas["SG_UF"].astype(str).str.strip()

# Filtrar registros da Paraíba
df_pb = df_taxas[df_taxas["SG_UF"] == "PB"].copy()

print("Total de registros no Brasil:", len(df_taxas))
print("Registros da Paraíba:", len(df_pb))

print("\nDistribuição por ano:")
display(df_pb["NU_ANO_CENSO"].value_counts(dropna=False))

print("\nDistribuição por dependência administrativa:")
display(df_pb["NO_DEPENDENCIA"].value_counts(dropna=False))

print("\nColunas das taxas do ensino médio:")
colunas_medio = [
    c for c in df_pb.columns
    if "CAT_MED" in c
]

print(colunas_medio)

print("\nAmostra das escolas da Paraíba:")
display(
    df_pb[
        [
            "NU_ANO_CENSO",
            "SG_UF",
            "CO_MUNICIPIO",
            "NO_MUNICIPIO",
            "CO_ENTIDADE",
            "NO_ENTIDADE",
            "NO_DEPENDENCIA",
            "1_CAT_MED",
            "2_CAT_MED",
            "3_CAT_MED"
        ]
    ].head(10)
)
```

```python

# CÉLULA 6 — Preparar a base de abandono do ensino médio público

# Manter escolas públicas da Paraíba
df_publicas = df_pb[
    df_pb["NO_DEPENDENCIA"].isin(["Municipal", "Estadual", "Federal"])
].copy()

# Converter os códigos de taxa para valores numéricos.
# Os símbolos "--" serão considerados ausentes, não como zero.
colunas_taxas = ["1_CAT_MED", "2_CAT_MED", "3_CAT_MED"]

for coluna in colunas_taxas:
    df_publicas[coluna] = pd.to_numeric(
        df_publicas[coluna].replace("--", pd.NA),
        errors="coerce"
    )

print("Escolas públicas na Paraíba:", len(df_publicas))

print("\nDisponibilidade das taxas do ensino médio:")
display(
    df_publicas[colunas_taxas].isna().sum()
    .rename("valores_ausentes")
    .to_frame()
    .assign(
        percentual_ausente=lambda x:
        (x["valores_ausentes"] / len(df_publicas) * 100).round(2)
    )
)

# Registros com taxa de abandono disponível
df_abandono = df_publicas[
    df_publicas["3_CAT_MED"].notna()
].copy()

print("\nRegistros públicos com taxa de abandono disponível:",
      len(df_abandono))

print("\nResumo da taxa de abandono:")
display(df_abandono["3_CAT_MED"].describe())

print("\nAmostra:")
display(
    df_abandono[
        [
            "NU_ANO_CENSO",
            "SG_UF",
            "CO_MUNICIPIO",
            "NO_MUNICIPIO",
            "CO_ENTIDADE",
            "NO_ENTIDADE",
            "NO_DEPENDENCIA",
            "3_CAT_MED"
        ]
    ].head(10)
)
```

```python

# CÉLULA 7 — Salvar a base de taxas da Paraíba

from pathlib import Path

# Selecionar os campos necessários
colunas = [
    "NU_ANO_CENSO",
    "SG_UF",
    "CO_MUNICIPIO",
    "NO_MUNICIPIO",
    "CO_ENTIDADE",
    "NO_ENTIDADE",
    "NO_DEPENDENCIA",
    "1_CAT_MED",
    "2_CAT_MED",
    "3_CAT_MED"
]

# Manter escolas públicas com taxa de abandono disponível
base_taxas_pb = df_abandono[colunas].copy()

# Padronizar nomes das colunas
base_taxas_pb = base_taxas_pb.rename(columns={
    "NU_ANO_CENSO": "ano",
    "SG_UF": "uf",
    "CO_MUNICIPIO": "codigo_municipio",
    "NO_MUNICIPIO": "municipio",
    "CO_ENTIDADE": "codigo_escola",
    "NO_ENTIDADE": "nome_escola",
    "NO_DEPENDENCIA": "dependencia",
    "1_CAT_MED": "taxa_aprovacao_medio",
    "2_CAT_MED": "taxa_reprovacao_medio",
    "3_CAT_MED": "taxa_abandono_medio"
})

# Salvar o resultado
pasta_saida = RAIZ / "dados" / "processado"
pasta_saida.mkdir(parents=True, exist_ok=True)

arquivo_saida = pasta_saida / "taxas_abandono_pb_2025.csv"

base_taxas_pb.to_csv(
    arquivo_saida,
    index=False,
    encoding="utf-8-sig"
)

print("Arquivo salvo:", arquivo_saida)
print("Total de registros:", len(base_taxas_pb))
print("Escolas únicas:", base_taxas_pb["codigo_escola"].nunique())

display(base_taxas_pb.head())
```
