```python

# CÉLULA 1 — Localizar e inspecionar os microdados de 2024

from pathlib import Path
import pandas as pd

RAIZ = Path.cwd()

if RAIZ.name == "notebooks":
    RAIZ = RAIZ.parent

arquivo_microdados = (
    RAIZ / "dados-brutos" / "microdados_ed_basica_2024.csv"
)

if not arquivo_microdados.exists():
    raise FileNotFoundError(
        f"Arquivo não encontrado: {arquivo_microdados}"
    )

print("Arquivo encontrado:", arquivo_microdados)
print("Tamanho (MB):", round(
    arquivo_microdados.stat().st_size / (1024 ** 2), 2
))

# Ler somente uma amostra para descobrir as colunas
df_amostra = pd.read_csv(
    arquivo_microdados,
    sep=";",
    encoding="latin1",
    nrows=5,
    low_memory=False
)

print("\nQuantidade de colunas:", len(df_amostra.columns))
print("\nNomes das colunas:")

for coluna in df_amostra.columns:
    print("-", coluna)

print("\nAmostra dos dados:")
display(df_amostra.head())
```

```python

# CÉLULA 2 — Selecionar dados relevantes da Paraíba

# Colunas de identificação e contexto escolar
colunas_desejadas = [
    "NU_ANO_CENSO",
    "SG_UF",
    "CO_MUNICIPIO",
    "NO_MUNICIPIO",
    "CO_ENTIDADE",
    "NO_ENTIDADE",
    "TP_DEPENDENCIA",
    "TP_LOCALIZACAO",
    "TP_ETAPA_ENSINO",
    "IN_INTERNET",
    "IN_BIBLIOTECA",
    "IN_LABORATORIO_INFORMATICA",
    "IN_BANDA_LARGA",
    "QT_MAT_MED",
    "QT_TUR_MED_INT"
]

# Manter apenas as colunas que realmente existem
colunas_disponiveis = [
    c for c in colunas_desejadas
    if c in df_amostra.columns
]

print("Colunas disponíveis na amostra:")
print(colunas_disponiveis)

# Ler somente as colunas disponíveis no arquivo completo
df_inep_pb = pd.read_csv(
    arquivo_microdados,
    sep=";",
    encoding="latin1",
    usecols=colunas_disponiveis,
    low_memory=False
)

# Filtrar Paraíba e ano de 2024
df_inep_pb = df_inep_pb[
    (df_inep_pb["SG_UF"] == "PB") &
    (df_inep_pb["NU_ANO_CENSO"] == 2024)
].copy()

print("\nDimensões após o filtro:", df_inep_pb.shape)
print("\nAmostra:")
display(df_inep_pb.head())
```

```python

# CÉLULA 3 — Identificar colunas de oferta e matrículas do ensino médio

# Ler somente o cabeçalho completo do arquivo
colunas_inep = pd.read_csv(
    arquivo_microdados,
    sep=";",
    encoding="latin1",
    nrows=0
).columns.tolist()

print("Colunas relacionadas a ensino médio:")
display([
    c for c in colunas_inep
    if "MED" in c.upper()
])

print("\nColunas relacionadas a etapa de ensino:")
display([
    c for c in colunas_inep
    if "ETAPA" in c.upper()
])

print("\nColunas relacionadas a matrícula:")
display([
    c for c in colunas_inep
    if "MAT" in c.upper()
])

print("\nColunas relacionadas a dependência:")
display([
    c for c in colunas_inep
    if "DEPENDENCIA" in c.upper()
])
```

```python

# CÉLULA 4 — Verificar oferta de ensino médio

colunas_necessarias = [
    "NU_ANO_CENSO",
    "SG_UF",
    "CO_MUNICIPIO",
    "CO_ENTIDADE",
    "TP_DEPENDENCIA",
    "TP_LOCALIZACAO",
    "IN_MED",
    "QT_MAT_MED",
    "QT_TUR_MED"
]

# Confirmar quais colunas existem
colunas_presentes = [
    c for c in colunas_necessarias
    if c in colunas_inep
]

print("Colunas encontradas:")
print(colunas_presentes)

# Ler somente as colunas necessárias
df_oferta = pd.read_csv(
    arquivo_microdados,
    sep=";",
    encoding="latin1",
    usecols=colunas_presentes,
    low_memory=False
)

# Filtrar Paraíba em 2024
df_oferta = df_oferta[
    (df_oferta["SG_UF"] == "PB") &
    (df_oferta["NU_ANO_CENSO"] == 2024)
].copy()

print("\nRegistros da Paraíba:", len(df_oferta))

print("\nDistribuição por dependência:")
display(
    df_oferta["TP_DEPENDENCIA"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nIndicador de oferta de ensino médio:")
if "IN_MED" in df_oferta.columns:
    display(df_oferta["IN_MED"].value_counts(dropna=False))

print("\nResumo das matrículas no ensino médio:")
if "QT_MAT_MED" in df_oferta.columns:
    display(df_oferta["QT_MAT_MED"].describe())

print("\nAmostra:")
display(df_oferta.head())
```

```python

# CÉLULA 5 — Filtrar escolas públicas com ensino médio

df_inep_medio = df_oferta[
    (df_oferta["TP_DEPENDENCIA"].isin([2, 3, 4])) &
    (df_oferta["IN_MED"] == 1) &
    (df_oferta["QT_MAT_MED"] > 0)
].copy()

print("Escolas públicas com ensino médio e matrículas positivas:",
      len(df_inep_medio))

print("\nDistribuição por dependência:")
display(
    df_inep_medio["TP_DEPENDENCIA"]
    .value_counts()
    .sort_index()
)

print("\nMatrículas no ensino médio:")
display(df_inep_medio["QT_MAT_MED"].describe())

print("\nCódigos de escola duplicados:")
print(df_inep_medio["CO_ENTIDADE"].duplicated().sum())

print("\nAmostra:")
display(df_inep_medio.head(10))
```

```python

# CÉLULA 6 — Integrar microdados de 2024 com taxas de 2025

# Carregar a base de taxas já processada
arquivo_taxas_pb = (
    RAIZ / "dados" / "processado" / "taxas_abandono_pb_2025.csv"
)

df_taxas_2025 = pd.read_csv(
    arquivo_taxas_pb,
    encoding="utf-8-sig"
)

# Padronizar os códigos para evitar problemas de tipo
df_inep_medio["CO_ENTIDADE"] = (
    pd.to_numeric(df_inep_medio["CO_ENTIDADE"], errors="coerce")
    .astype("Int64")
)

df_taxas_2025["codigo_escola"] = (
    pd.to_numeric(df_taxas_2025["codigo_escola"], errors="coerce")
    .astype("Int64")
)

# Verificar duplicatas antes da junção
print("Duplicatas na base de 2024:",
      df_inep_medio["CO_ENTIDADE"].duplicated().sum())

print("Duplicatas na base de taxas de 2025:",
      df_taxas_2025["codigo_escola"].duplicated().sum())

# Selecionar variáveis explicativas de 2024
colunas_2024 = [
    "CO_ENTIDADE",
    "CO_MUNICIPIO",
    "TP_DEPENDENCIA",
    "TP_LOCALIZACAO",
    "QT_MAT_MED",
    "QT_TUR_MED"
]

base_2024 = df_inep_medio[colunas_2024].copy()

# Integrar: manter todas as escolas de 2024, mesmo sem taxa em 2025
base_integrada = base_2024.merge(
    df_taxas_2025[
        ["codigo_escola", "taxa_abandono_medio"]
    ],
    left_on="CO_ENTIDADE",
    right_on="codigo_escola",
    how="left",
    validate="one_to_one"
)

print("\nEscolas na base de 2024:", len(base_2024))
print("Escolas após a integração:", len(base_integrada))
print("Com taxa de abandono em 2025:",
      base_integrada["taxa_abandono_medio"].notna().sum())
print("Sem taxa de abandono em 2025:",
      base_integrada["taxa_abandono_medio"].isna().sum())

print("\nAmostra da base integrada:")
display(base_integrada.head(10))
```

```python

# CÉLULA 7 — Auditar a base integrada e preparar a base analítica

# Separar registros com e sem taxa de abandono
base_modelagem = base_integrada[
    base_integrada["taxa_abandono_medio"].notna()
].copy()

base_sem_taxa = base_integrada[
    base_integrada["taxa_abandono_medio"].isna()
].copy()

print("=== COBERTURA DA INTEGRAÇÃO ===")
print("Total de escolas:", len(base_integrada))
print("Com taxa:", len(base_modelagem))
print("Sem taxa:", len(base_sem_taxa))
print(
    "Cobertura (%):",
    round(len(base_modelagem) / len(base_integrada) * 100, 2)
)

print("\n=== DISTRIBUIÇÃO DA TAXA DE ABANDONO ===")
display(base_modelagem["taxa_abandono_medio"].describe())

print("\n=== TAXA POR DEPENDÊNCIA ADMINISTRATIVA ===")
display(
    base_modelagem.groupby("TP_DEPENDENCIA")[
        "taxa_abandono_medio"
    ].agg(["count", "mean", "median", "std"]).round(2)
)

print("\n=== CORRELAÇÕES INICIAIS ===")
display(
    base_modelagem[
        ["QT_MAT_MED", "QT_TUR_MED", "taxa_abandono_medio"]
    ].corr().round(3)
)

print("\n=== AMOSTRA DA BASE DE MODELAGEM ===")
display(base_modelagem.head(10))
```

```python

# CÉLULA 8 — Salvar a base analítica

pasta_saida = RAIZ / "dados" / "processado"
pasta_saida.mkdir(parents=True, exist_ok=True)

# Base integrada completa, inclusive escolas sem taxa
arquivo_completo = pasta_saida / "base_integrada_pb_2024_2025.csv"

base_integrada.to_csv(
    arquivo_completo,
    index=False,
    encoding="utf-8-sig"
)

# Base com alvo disponível para análises exploratórias
arquivo_modelagem = pasta_saida / "base_modelagem_pb_2025.csv"

base_modelagem.to_csv(
    arquivo_modelagem,
    index=False,
    encoding="utf-8-sig"
)

print("Base integrada salva em:", arquivo_completo)
print("Base com alvo disponível salva em:", arquivo_modelagem)
print("Linhas da base integrada:", len(base_integrada))
print("Linhas com taxa disponível:", len(base_modelagem))

print("\nValores ausentes por coluna:")
display(base_modelagem.isna().sum().to_frame("ausentes"))

print("\nAmostra final:")
display(base_modelagem.head())
```
